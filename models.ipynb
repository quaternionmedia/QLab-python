```python
from sqlmodel import create_engine, SQLModel, Session, select
from qlab import models
from rich import print

database = 'sqlite:///mix/seuss3.tmix'
# database = 'sqlite:///mix/SheKillsMonsters.sqlite'

engine = create_engine(database) #, echo=True)
```

```python
SQLModel.metadata.create_all(engine)
```

```python
with Session(engine) as session:
    cue = models.Cue(name="Q2", number=10, point=2)
    session.add(cue)
    session.commit()
```

```python
with Session(engine) as session:
    cues = session.exec(select(models.Cue).order_by(models.Cue.number, models.Cue.point)).first() # .all()
    # for cue in cues:
    #     print(cue)
    print(cues)
```
