```python
import csv
from qlab.models import Profile
from faker import Faker
fake = Faker()

db_path = 'mix/seuss9.tmix'


with open('config/characters.csv', 'r') as f:
    reader = csv.DictReader(f)
    characters = [Profile(name=row['name'], channel=int(row['channel'])) for row in reader]
# characters
```

```python
from sqlmodel import create_engine, SQLModel, Session, select
from qlab import models
from rich import print


engine = create_engine(f'sqlite:///{db_path}') #, echo=True)
SQLModel.metadata.create_all(engine)
```

```python
with Session(engine) as session:
    for character in characters:
        profile = session.get(Profile, character.channel)
        profile.name = character.name
        session.add(profile)
    session.commit()
```

```python
from qlab.models import Ensemble

with open('config/ensemble.csv', 'r') as f:
    reader = csv.DictReader(f)
    ensembles = [Ensemble(name=row['name'], channels=None) for row in reader]
# ensembles
```

```python
with Session(engine) as session:
    session.add_all(ensembles)
    session.commit()
```
